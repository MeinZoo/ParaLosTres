## Задача 1. Первая функция и `return`

*Новое: `def`, параметр, `return`, `None`*

Функцию создают при помощи ключевого слова `def`: имя, параметры в скобках, двоеточие, тело с отступом.
Функция, которая что-то **считает**, должна **возвращать** результат через `return`,
а не печатать его. Функция без `return` возвращает `None` - «ничего».

````python
def count_chars(text):
    return len(text)

def show_chars(text):
    print(len(text))

a = count_chars("кот спит")
b = show_chars("кот спит")      # напечатает 8, но вернёт None
print("a =", a, " b =", b)
print(count_chars("кот") + count_chars("пёс"))   # с результатом можно считать
````

**Ваша очередь:** напишите функцию `word_count(text)`, которая **возвращает** число слов в тексте
(слова разделены пробелами). Напечатайте число слов в первом заголовке,
а потом циклом посчитайте, сколько всего слов во всех текстах `TEXTS`.

**Должно получиться:**

```
4 | Сборная вышла в четвертьфинал
всего слов в ленте: 256
```

````python
# Задача 1

def word_count(text):
    return len(text.split())

print(word_count(TITLES[0]), "|", TITLES[0])
total = 0
for text in TEXTS:
    total += word_count(text)
print("всего слов в ленте:", total)
````
